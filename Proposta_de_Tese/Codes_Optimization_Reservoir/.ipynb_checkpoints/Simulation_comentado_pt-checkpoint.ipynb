```
# -*- coding: utf-8 -*-
"""
    Caso utilize este código, cite o trabalho relacionado:
       https://doi.org/10.1016/j.geoen.2024.213621
"""

# Importações de bibliotecas necessárias
import numpy as np
import re
import pandas as pd
import numpy as np
import random
import time
import os
import matplotlib.pyplot as plt
import shutil

# Define o diretório principal do projeto
MAINDIR = r"/Users/calva/OneDrive/Documentos/WellingtonCodesOptimizationReservoirModels"

# Define o diretório temporário para onde arquivos de simulação serão copiados
TempFolder = os.path.join(MAINDIR, "Temp")
# Define o diretório do modelo (Egg - Two ou Egg - Four)
ModelDir = os.path.join(MAINDIR, "Egg - Two") # Altere para "Egg - Four" para o segundo período

# Define o diretório do caso de simulação (deve conter códigos e arquivos necessários)
Case_path = os.path.join(MAINDIR, "Egg - Two") # Altere para "Egg - Four" para o segundo período

# Cria a pasta temporária se ela não existir
if not os.path.exists(TempFolder):
    os.mkdir(TempFolder)

# Lê linhas específicas do arquivo ACTIVE.INC do modelo (linhas 2 a 421)
with open(os.path.join(ModelDir, "ACTIVE.INC"), "r") as file:
    c = file.readlines()[2:422]

# Inicializa variáveis para manipulação do array de atividade
k = 0
i = 0
act = np.zeros((60,60,7))
for line in c:
    if i == 60:
        k += 1
        i = 0
    if line != "\n":
        line.strip("\n")
        act[i,:,k] = np.array([int(x) for x in line.split()])
        i += 1

# Função para modificar arquivos do caso de simulação (ajusta posições dos poços e realizações)
def modify_file(File_path, new_value, realization):
    try:
        # Remove arquivo EGG.RSM se já existir no diretório temporário
        if os.path.exists(os.path.join(TempFolder, "EGG.RSM")):
            os.remove(os.path.join(TempFolder, "EGG.RSM"))
    except:
        pass   
    # Copia todos os arquivos do caso para a pasta temporária
    shutil.copytree(File_path, TempFolder, dirs_exist_ok = True)
    os.chdir(TempFolder)
    # Lê o arquivo EGG.DATA
    with open("EGG.DATA", "r") as file:
        content = file.readlines()
    # Modifica as linhas necessárias para atualizar as realizações e coordenadas dos poços
    for c, line in enumerate(content):
        if "Realizations" in line:
            content[c] = f"'../Realizations/{realization}'"
        if "WELSPECS" in line:
            for i in range(int(len(new_value)/2)):
                tokens = content[c+i+1].split()
                tokens[2] = str(int(new_value[i*2])+1)
                tokens[3] = str(int(new_value[i*2+1])+1)
                content[c+i+1] = " ".join(tokens)+"\n"
            break
    # Salva as modificações no arquivo EGG.DATA
    with open("EGG.DATA", "w") as file:
        file.writelines(content)

# Função para extrair um valor de um arquivo, dado a linha e a coluna
def extract_value_from_file(file_path, target_line, target_column):
    try:
        with open(file_path, 'r') as file:
            # Lê todas as linhas do arquivo
            lines = file.readlines()
            # Retorna o valor desejado convertido para float
            return float(lines[target_line].split()[target_column])
    except FileNotFoundError:
        print(f"Erro: Arquivo não encontrado em {file_path}.")
    except Exception as e:
        print(f"Ocorreu um erro: {e}")

# Função para gerar um ponto aleatório dentro da máscara de atividade (leaf)
def generate_random_point(width, height):
    x = np.random.randint(0, width)
    y = np.random.randint(0, height)
    while not np.any(act[x,y]):
        x = np.random.randint(0, width)
        y = np.random.randint(0, height)
        # Se houver uma máscara de folha, pode ser verificado aqui
    return x, y

# Define o tamanho do grid
width, height = 60, 60        
# Testa a função de geração de ponto aleatório
generate_random_point(width, height)

# Número de dados a serem gerados
num_data = 2500
# Nomes das colunas do DataFrame final
cols = ['PRODX1', 'PRODY1','INJX1','INJY1','COP']
# Lista para armazenar os resultados
lst = []
# Lista de realizações utilizadas no estudo
Realizations = ["PERM60_ECL.INC", "PERM76_ECL.INC", "PERM82_ECL.INC"]

# Loop sobre todas as realizações
for i in range(len(Realizations)):
    # Gera coordenadas aleatórias para poços produtores e injetores
    pointsp = [generate_random_point(width, height) for _ in range(num_data)]
    random.shuffle(pointsp)
    pointsi = [generate_random_point(width, height) for _ in range(num_data)]
    data_list1 = []
    # Monta o DataFrame com todas as combinações
    for k in range(0,num_data):
        new_row = {
            'PX1': pointsp[k][0],
            'PY1': pointsp[k][1],
            'IX1': pointsi[k][0],
            'IY1': pointsi[k][1]
        }
        data_list1.append(new_row)
    data_list1=pd.DataFrame(data_list1)
    lst1=[]
    # Para cada combinação de coordenadas, modifica arquivos, executa simulação e extrai resultado
    for e in range(len(data_list1)):
        # Atualiza o arquivo do caso com novas posições e realização
        modify_file(Case_path, data_list1.iloc[e].values, Realizations[i]) # X,Y do injetor1

        # Executa o simulador externo via comando de sistema (Windows BAT)
        os.system("File_Run.bat")
        # Aguarda o arquivo de resultado ser gerado
        while not os.path.exists(TempFolder+"\\EGG.RSM"):
            continue
        time.sleep(1)
        # Extrai o valor de óleo acumulado do arquivo de saída
        Cum_Oil = extract_value_from_file(TempFolder+"\\EGG.RSM", -1, 2)  # Extrai o valor desejado
        
        # Armazena os dados (coordenadas, realização e óleo acumulado) na lista de resultados
        lst.append((data_list1.iloc[e].values[0],data_list1.iloc[e].values[1],data_list1.iloc[e].values[2],data_list1.iloc[e].values[3],Realizations[i], Cum_Oil))
        os.chdir(MAINDIR)
    # lst.append(lst1)  # Linha comentada

# Converte os resultados para DataFrame do pandas
df = pd.DataFrame(lst)
# Salva o DataFrame em um arquivo Excel
df.to_excel("First.xlsx") 
```
