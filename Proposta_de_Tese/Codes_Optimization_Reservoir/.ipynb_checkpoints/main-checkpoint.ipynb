```python
import subprocess
import os

# caminho da pasta onde estão os scripts
base_dir = "/mnt/c/Users/calva/OneDrive/Documentos/Doutorado/Proposta_de_Tese/Codes_Optimization_Reservoir/"

scripts = ["Optimization_1.py", "Optimization_2.py"]

for script in scripts:
    print(f"\n>>> Rodando {script} no WSL (Windows Subsystem for Linux)...")
    result = subprocess.run(
        ["wsl", "python3", os.path.join(base_dir, script)],
        capture_output=True,
        text=True
    )
    
    # mostra a saída do script no notebook
    print(result.stdout)
    
    # se der erro, mostra no terminal
    if result.stderr:
        print("⚠️ Erro:", result.stderr)

print("\n>>> Finalizado com sucesso!")
```
