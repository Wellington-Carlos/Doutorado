```python
import numpy as np
import matplotlib.pyplot as plt

# Funções de ativação
def step_function(x):
    return np.where(x >= 0, 1, 0)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def softmax(x):
    e_x = np.exp(x - np.max(x))
    return e_x / e_x.sum(axis=0)

# Intervalo de entrada
x = np.linspace(-10, 10, 200)

# Criação da figura 2x2
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# (a) Degrau
axs[0, 0].plot(x, step_function(x), color='black')
axs[0, 0].set_title('(a)')
axs[0, 0].grid(True, alpha=0.2)

# (b) Sigmoide
axs[0, 1].plot(x, sigmoid(x), color='blue')
axs[0, 1].set_title('(b)')
axs[0, 1].grid(True, alpha=0.2)

# (c) ReLU
axs[1, 0].plot(x, relu(x), color='green')
axs[1, 0].set_title('(c)')
axs[1, 0].grid(True, alpha=0.2)

# (d) Softmax com duas e três classes
X_soft2 = np.vstack([x, -x])
Y_soft2 = softmax(X_soft2)
axs[1, 1].plot(x, Y_soft2[0, :], 'r')

axs[1, 1].set_title('(d)')
axs[1, 1].grid(True, alpha=0.2)

# Ajustes finais
for ax in axs.flat:
    ax.set_xlabel('v')
    #ax.set_ylabel('Output')

plt.tight_layout()

output_path = r"C:\Users\calva\OneDrive\Documentos\Doutorado\Proposta_de_Tese\wellington\images\funcoes_ativacao.png"
plt.savefig(output_path, dpi=400, bbox_inches='tight')
plt.show()
```
