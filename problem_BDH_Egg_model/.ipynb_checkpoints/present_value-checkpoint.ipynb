```python
def present_value(df, taxa_desconto):
    """
    Calcula o Valor Presente considerando um deslocamento no tempo.
    shift = 0 → t=0 é descontado com expoente 0.
    shift = 1 → t=0 é descontado como se fosse um ano no futuro.
    """
    num_periodos = df.shape[1]
    fatores_desconto = 1 / (1 + taxa_desconto) ** (np.arange(num_periodos))
    valor_presente = df.values @ fatores_desconto
    return pd.Series(valor_presente, index=df.index)
```

```python
pv_t_minus_1 = present_value(df1, dr)

print(f'Value of the initial investment: ${init_invest:.2f}')
print(f'PV of cash flows: ${np.mean(pv_t_minus_1):.2f}')
print(f'Value of the project without options (NPV): ${np.mean(pv_t_minus_1) - init_invest:.2f}')
```
