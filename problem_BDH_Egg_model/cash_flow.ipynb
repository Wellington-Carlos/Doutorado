```python
def cash_flow_loops(trialno, N, frame, prices_int_all, VOC_price_int_all, fixed_cost, profit):
    # Seleciona colunas numéricas de anos (0..N), ignora "PV"
    year_cols = [c for c in frame.columns if str(c).isdigit()]
    year_cols = sorted(year_cols, key=lambda x: int(x))  # garante ordem 0..N

    # Pega linha 'Production level' como array de float
    prod = pd.to_numeric(frame.loc['Production level', year_cols], errors='coerce').to_numpy(dtype=float)

    cash = np.zeros((trialno, N+1), dtype=float)

    for i in range(trialno):
        for j in range(N+1):  # agora inclui t=0 até t=N
            cash[i, j] = ((prod[j] * float(prices_int_all[i, j]))
                          - (prod[j] * float(VOC_price_int_all[i, j]) + float(fixed_cost))) \
                         * (1.0 - float(profit))
    return cash
```

```python
# Calcular cash flow
cash_arr = cash_flow_loops(trialno, N, frame, prices_int_all, VOC_price_int_all, fixed_cost, profit)

row_header1 = [f"Path {i}" for i in range(1, trialno + 1)]
df1 = pd.DataFrame(cash_arr, index=row_header1, columns=[str(c) for c in range(N+1)])

# Criar resumo (3 primeiras, reticências, 3 últimas)
df1_summary = pd.concat([
    df1.head(3),
    pd.DataFrame([['...' for _ in df1.columns]], columns=df1.columns),
    df1.tail(3)
])

print(df1_summary)
```
